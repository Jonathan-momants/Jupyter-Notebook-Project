# Momants language detection

Detect Dutch, English, and German visitor messages with Lingua. Weak or unusable evidence is returned as `und`.

## 1. Import the language module

```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

import momants_language
```

## 2. Choose input and output paths

```python
CSV_PATH = PROJECT_ROOT / "data" / "tests" / "conversations_100.csv"
OUTPUT_DIRECTORY = PROJECT_ROOT / "results"
```

## 3. Inspect per-message language

The shared loader keeps only the six permitted fields and applies PII masking before language detection.

```python
data = momants_language.load_momants_csv(CSV_PATH)
visitors = momants_language.select_visitor_messages(data)
message_languages = momants_language.classify_messages(visitors)
message_languages[["conversation_id", "created_at", "text", "language", "confidence"]].head(20)
```

## 4. Build the customer-facing conversation statistic

Conversation language is weighted by message character count. `und` messages do not contribute to the weighting.

```python
conversation_languages = momants_language.process_csv(
    CSV_PATH,
    output_directory=OUTPUT_DIRECTORY,
)
conversation_languages.head(20)
```
