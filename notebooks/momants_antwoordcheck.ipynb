# Momants Answer Check

This notebook determines whether visitor questions have been answered in each conversation. It does not determine sentiment, topic, or urgency. Only the permitted Momants fields are loaded.

## 1. Import the answer check

```python
from pathlib import Path
import sys

PROJECT_DIR = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
if str(PROJECT_DIR) not in sys.path:
    sys.path.insert(0, str(PROJECT_DIR))

from momants_antwoordcheck import detect_questions, process_csv
from momants_sentiment import load_momants_csv
```

## 2. Select the CSV file and output directory

```python
CSV_PATH = PROJECT_DIR / "attached_assets" / "test_gesprekken_100.csv"
OUTPUT_DIR = PROJECT_DIR / "results"

print(f"Input: {CSV_PATH}")
print(f"Output directory: {OUTPUT_DIR}")
```

## 3. Check question detection

This step does not start the NLI model and shows counts only.

```python
data = load_momants_csv(CSV_PATH)
questions = detect_questions(data)

print(f"Message rows: {len(data)}")
print(f"Detected questions: {len(questions)}")
print(f"Conversations with questions: {questions['conversation_id'].nunique()}")
```

## 4. Determine whether questions were answered

This step loads the multilingual NLI model and writes a file with its start time in its name, for example `results/answer_check_per_conversation_20260901_143522_123456.csv`.

```python
answer_check_results = process_csv(
    csv_path=CSV_PATH,
    output_dir=OUTPUT_DIR,
    batch_size=32,
)

print(f"Complete: processed {len(answer_check_results)} conversations.")
print(f"File: {answer_check_results.attrs['output_path']}")
answer_check_results.head(10)
```
