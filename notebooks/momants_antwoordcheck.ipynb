# Momants antwoordcheck

Deze notebook bepaalt per gesprek of vragen van bezoekers zijn beantwoord. Er wordt geen sentiment, onderwerp of urgentie bepaald. Alleen de toegestane Momants-velden worden ingelezen.

## 1. Importeer de antwoordcheck

```python
from pathlib import Path
import sys

PROJECTMAP = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
if str(PROJECTMAP) not in sys.path:
    sys.path.insert(0, str(PROJECTMAP))

from momants_antwoordcheck import detecteer_vragen, verwerk_csv
from momants_sentiment import laad_momants_csv
```

## 2. Kies de CSV en uitvoermap

```python
CSV_PAD = PROJECTMAP / "notebooks" / "public_main_membermessage_export_2026-08-31_150601.csv"
UITVOERMAP = PROJECTMAP / "resultaten"

print(f"Invoer: {CSV_PAD}")
print(f"Uitvoer: {UITVOERMAP}")
```

## 3. Controleer de vraagdetectie

Deze stap start het NLI-model nog niet en toont uitsluitend aantallen.

```python
data = laad_momants_csv(CSV_PAD)
vragen = detecteer_vragen(data)

print(f"Berichtrijen: {len(data)}")
print(f"Gedetecteerde vragen: {len(vragen)}")
print(f"Gesprekken met vragen: {vragen['conversation_id'].nunique()}")
```

## 4. Bepaal of de vragen beantwoord zijn

Deze stap laadt het meertalige NLI-model en schrijft `resultaten/antwoordcheck_per_gesprek.csv`. Het resultaat bevat geen oorspronkelijke berichttekst.

```python
antwoordresultaten = verwerk_csv(
    csv_pad=CSV_PAD,
    uitvoermap=UITVOERMAP,
    batchgrootte=32,
)

print(f"Klaar: {len(antwoordresultaten)} gesprekken verwerkt.")
print(f"Bestand: {UITVOERMAP / 'antwoordcheck_per_gesprek.csv'}")
antwoordresultaten.head(10)
```
